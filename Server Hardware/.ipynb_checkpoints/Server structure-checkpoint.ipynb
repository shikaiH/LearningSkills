## Server structure

1. Rack (cabinet)
   - **A. Rack Architecture Comparison**

| Spec definition | EIA-310 | Open compute project starts from 2022<br>(Meta, Google, Microsoft...etc) | NVIDIA starts from 2023 continuous updates |
| :--- | :--- | :--- | :--- |
| Rack type | Standard Rack | OCP open rack | NVIDIA MGX |
| internal mounting width | 19" inches<br>(482.6mm) | 21" inches<br>(537mm) | 19" inches and keep the OCP rack(ORv3) advantages <br> e.g. busbar and rack manifold |
| Height Unit | 1U, 44.45mm<br>Total unit usually 42U | 1U 48mm<br>Total unit usually 44OU | U or MGX proprietary form factor (Supports 1U/2U/4U modular nodes) |
| Power delivery | Individual power cord per chassis (AC PDU) | Rear centralized 48V DC Busbar | Rear centralized 48V / 5000A ultra-high current |
| Thermal design | air cooler, front to back air flow | Air cooling, open liquid cooling manifold | Mandatory blind-mate DLC (Direct-to-Chip Liquid Cooling), with manifolds integrated directly inside the rack |
| Rack backplane | manual network cabling | power busbar and liquid pipes only | Proprietary NVLink Cable Cartridge (ultra-high-speed copper backplane)<br>connecting 72 GPUs into "a single giant GPU" |
| Single-rack power density | 10 kW – 30 kW | 30 kW – 70 kW | **120 kW – 250 kW** (MGX 1.2 specification) |
