1. Rack (cabinet)
    1. Standard Rack(EIA-310)
       1. Internal Mounting width: 19 inches(482.6mm)
       2. Height Unit: 44.45mm
       3. Power delivery: Individual power cords per chassis (AC PDU)
       4. Thermal design: Primarily air-cooled (front to back air floor)
       5. Rack backplane: None (manual network cabling)
       6. Single-rack power density
